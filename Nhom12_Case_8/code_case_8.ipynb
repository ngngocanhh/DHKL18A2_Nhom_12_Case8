1. ĐỌC FILE CSV

```python
# ĐỌC 2 FILE CSV
import numpy as np
store_a = np.genfromtxt("store_a.csv", delimiter=",", dtype=str, encoding="utf-8")
store_b = np.genfromtxt("store_b.csv", delimiter=",", dtype=str, encoding="utf-8")
```

2. TÁCH DATA VÀ SALES THÀNH CÁC MẢNG RIÊNG

```python
# Store A
header_a = store_a[0]   # lấy dòng đầu tiên trong mảng, đó là header(tiêu đề) ["date", "sales"])
data_a = store_a[1:]    # lấy tất cả các dòng còn lại, tức là dữ liệu thực sự (không có header)

dates_a = data_a[:, 0]  # tất cả các dòng (:), cột thứ 0 → cột date
sales_a = data_a[:, 1].astype(float) # tất cả các dòng, cột thứ 1 → cột sales
# .astype(float) → chuyển kiểu dữ liệu sang số thực, vì khi đọc CSV bằng dtype=str, tất cả dữ liệu đều là chuỗi (string).
```

```python
# Store B
header_b = store_b[0]
data_b = store_b[1:]

dates_b = data_b[:, 0]
sales_b = data_b[:, 1].astype(float)
```

3. CHUYỂN DATE SANG DATETIME64

```python
dates_a_dt = dates_a.astype('datetime64[D]')
dates_b_dt = dates_b.astype('datetime64[D]')
# 'datetime64[D]' → đại diện ngày (day-level)
# astype('datetime64[D]') → chuyển chuỗi có định dạng 'YYYY-MM-DD' sang datetime.
```

4.1 KIỂM TRA DTYPE, SHAPE

```python
# STORE A
print("Dates dtype:", dates_a_dt.dtype)
print("Dates shape:", dates_a_dt.shape)

print("Sales dtype:", sales_a.dtype)
print("Sales shape:", sales_a.shape)
```

```python
# STORE B
print("Dates dtype:", dates_b_dt.dtype)
print("Dates shape:", dates_b_dt.shape)

print("Sales dtype:", sales_b.dtype)
print("Sales shape:", sales_b.shape)
```

4.2 IN 5 GIÁ TRỊ ĐẦU/CUỐI

```python
# STORE A
print("5 giá trị đầu của dates:", dates_a_dt[:5])
print("5 giá trị cuối của dates:", dates_a_dt[-5:])

print("5 giá trị đầu của sales:", sales_a[:5])
print("5 giá trị cuối của sales:", sales_a[-5:])
```

```python
# STORE B
print("5 giá trị đầu của dates:", dates_b_dt[:5])
print("5 giá trị cuối của dates:", dates_b_dt[-5:])

print("5 giá trị đầu của sales:", sales_b[:5])
print("5 giá trị cuối của sales:", sales_b[-5:])
```

5. SLICING 30 NGÀY ĐẦU/GIỮA/CUỐI

```python
first_30_a = sales_a[:30]
mid_30_a   = sales_a[30:60]
last_30_a  = sales_a[60:90]

first_30_b = sales_b[:30]
mid_30_b   = sales_b[30:60]
last_30_b  = sales_b[60:90]
```

6. BOOLEAN MASK

```python
mean_a = np.mean(sales_a)
mean_b = np.mean(sales_b)

mask_high_a = sales_a > 1.5 * mean_a
mask_high_b = sales_b > 1.5 * mean_b

mask_low_a = sales_a < 0.5 * mean_a
mask_low_b = sales_b < 0.5 * mean_b

sales_a_high = sales_a[mask_high_a]
sales_a_low  = sales_a[mask_low_a]

sales_b_high = sales_b[mask_high_b]
sales_b_low  = sales_b[mask_low_b]

dates_a_high = dates_a[mask_high_a]
dates_a_low  = dates_a[mask_low_a]
```

7. BROADCASTING CHÊNH LỆCH DOANH SỐ

 8. FANCY INDEXING

STORE A

```python
# Sắp xếp chỉ số theo doanh số tăng dần
sorted_a_idx = np.argsort(sales_a)    # np.argsort(sales_a) → lấy chỉ số sắp xếp doanh số tăng dần.

# Bottom-10 doanh số thấp nhất
bottom_10_idx_a = sorted_a_idx[:10]
bottom_10_sales_a = sales_a[bottom_10_idx_a]
bottom_10_dates_a_dt = dates_a_dt[bottom_10_idx_a]

# Top-10 doanh số cao nhất
top_10_idx_a = sorted_a_idx[-10:][::-1]  # đảo ngược từ cao xuống thấp
top_10_sales_a = sales_a[top_10_idx_a]
top_10_dates_a_dt = dates_a_dt[top_10_idx_a]

# In kết quả
print("Top-10 doanh số cao nhất STORE A kèm date:")
for date_a, sale_a in zip(top_10_dates_a_dt, top_10_sales_a):
    print(date_a, sale_a)

print("\nTop-10 doanh số thấp nhất STORE A kèm date:")
for date_a, sale_a in zip(bottom_10_dates_a_dt, bottom_10_sales_a):    # zip() ghép các chỉ số
    print(date_a, sale_a)
```

STORE B

```python
# Sắp xếp chỉ số theo doanh số tăng dần
sorted_b_idx = np.argsort(sales_b)

# Bottom-10 doanh số thấp nhất
bottom_10_idx_b = sorted_b_idx[:10]
bottom_10_sales_b = sales_b[bottom_10_idx_b]
bottom_10_dates_b_dt = dates_b_dt[bottom_10_idx_b]

# Top-10 doanh số cao nhất
top_10_idx_b = sorted_b_idx[-10:][::-1]  # đảo ngược từ cao xuống thấp
top_10_sales_b = sales_b[top_10_idx_b]
top_10_dates_b_dt = dates_b_dt[top_10_idx_b]

# In kết quả
print("Top-10 doanh số cao nhất STORE B kèm date:")
for date_b, sale_b in zip(top_10_dates_b_dt, top_10_sales_b):
    print(date_b, sale_b)

print("\nTop-10 doanh số thấp nhất STORE B kèm date:")
for date_b, sale_b in zip(bottom_10_dates_b_dt, bottom_10_sales_b):    # zip() ghép các chỉ số
    print(date_b, sale_b)
```

9. PHÂN TÍCH BIẾN ĐỘNG DOANH SỐ

9.1 Chênh lệch doanh số giữa các ngày liên tiếp

```python
diff_a = np.diff(sales_a)       # biến động từng ngày của Store A
diff_b = np.diff(sales_b)       # biến động từng ngày của Store B
```

9.2 Biến động mạnh

```python
threshold_a = 2 * np.std(diff_a)        # Đặt ngưỡng biến động mạnh cho Store A là 2 lần độ lệch chuẩn của chênh lệch doanh số hàng ngày
threshold_b = 2 * np.std(diff_b)        # Đặt ngưỡng biến động mạnh cho Store B là 2 lần độ lệch chuẩn của chênh lệch doanh số hàng ngày
```

9.3 Tìm vị trí có biến động mạnh 


```python
strong_move_a = np.where(np.abs(diff_a) > threshold_a)[0]       # np.abs(diff_a) > threshold_a → True ở những ngày biến động lớn bất thường
strong_move_b = np.where(np.abs(diff_b) > threshold_b)[0]       # np.where(...)[0] trả về chỉ số của những ngày đó
```

9.4 Các ngày mà doanh số tăng/giảm bất thường 

```python
print("Ngày biến động mạnh Store A:", strong_move_a)        # In ra các chỉ số ngày có biến động doanh số bất thường ở Store A
print("Ngày biến động mạnh Store B:", strong_move_b)        # In ra các chỉ số ngày có biến động doanh số bất thường ở Store B
```

10. THỐNG KÊ TỔNG HỢP ( MEAN, STD, MEDIAN, MIN, MAX)

STORE A

```python
mean_a = np.mean(sales_a)       # Tính trung bình doanh số Store A
std_a = np.std(sales_a)         # Tính độ lệch chuẩn Store A
min_a = np.min(sales_a)         # Tìm doanh số thấp nhất Store A
max_a = np.max(sales_a)         # Tìm doanh số cao nhất Store A
median_a = np.median(sales_a)   # Tìm trung vị doanh số Store A
print("Store A -> mean:", mean_a, "std:", std_a, "min:", min_a, "max:", max_a, "median:", median_a)     # In thống kê Store A
```

STORE B

```python
mean_b = np.mean(sales_b)       # Tính trung bình doanh số Store B
std_b = np.std(sales_b)         # Tính độ lệch chuẩn Store B
min_b = np.min(sales_b)         # Tìm doanh số thấp nhất Store B
max_b = np.max(sales_b)         # Tìm doanh số cao nhất Store B
median_b = np.median(sales_b)   # Tìm trung vị doanh số Store B
print("Store B -> mean:", mean_b, "std:", std_b, "min:", min_b, "max:", max_b, "median:", median_b)     # In thống kê Store B
```
